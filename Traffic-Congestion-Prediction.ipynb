```python
# ============================================================
# 🚦 TRAFFIC CONGESTION PREDICTION - ML MODEL
# ============================================================

# Step 1: Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import joblib


# ============================================================
# Step 2: Load Dataset
# ============================================================

df = pd.read_excel(
    r"D:\Traffic_Prediction_48000_Featured.xlsx"
)

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# Step 3: Dataset Information
# ============================================================

print("\nDataset Information:")
df.info()


# ============================================================
# Step 4: Check Missing Values
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# Step 5: Check Columns
# ============================================================

print("\nAvailable Columns:")
print(df.columns.tolist())


# ============================================================
# Step 5.1: Convert DayOfWeek to Numbers
# ============================================================

day_mapping = {
    "Monday": 0,
    "Tuesday": 1,
    "Wednesday": 2,
    "Thursday": 3,
    "Friday": 4,
    "Saturday": 5,
    "Sunday": 6
}

df["DayOfWeek"] = df["DayOfWeek"].map(day_mapping)


# ============================================================
# Step 6: Select Features and Target
# ============================================================

features = [
    "Hour",
    "Junction",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "Weekend"
]

X = df[features]

y = df["Vehicles"]


print("\nFeatures Used:")
print(features)

print("\nTarget:")
print("Vehicles")


# ============================================================
# Step 7: Check Feature Data
# ============================================================

print("\nX Shape:", X.shape)
print("Y Shape:", y.shape)

print("\nX Sample:")
print(X.head())

print("\nY Sample:")
print(y.head())


# ============================================================
# Step 8: Train-Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)


# ============================================================
# Step 9: Create Random Forest Model
# ============================================================

from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=50,
    max_depth=10,
    random_state=42,
    n_jobs=1
)


# ============================================================
# Step 10: Train Model
# ============================================================

model.fit(X_train, y_train)

print("Model Training Completed!")


# ============================================================
# Step 11: Prediction
# ============================================================

y_pred = model.predict(X_test)

print("\nFirst 10 Predictions:")
print(y_pred[:10])


# ============================================================
# Step 12: Model Evaluation
# ============================================================

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)


print("\n====================================")
print("MODEL PERFORMANCE")
print("====================================")

print("MAE      :", mae)
print("MSE      :", mse)
print("RMSE     :", rmse)
print("R2 Score :", r2)


# ============================================================
# Step 13: Actual vs Predicted
# ============================================================

result = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print("\nActual vs Predicted:")
print(result.head(10))


# ============================================================
# Step 14: Vehicle Statistics
# ============================================================

print("\n====================================")
print("VEHICLE STATISTICS")
print("====================================")

print(df["Vehicles"].describe())


# ============================================================
# Step 15: Junction Information
# ============================================================

print("\nAvailable Junctions:")

print(df["Junction"].unique())


# ============================================================
# Step 16: Congestion Level Function
# ============================================================

def get_congestion_level(vehicles):

    if vehicles < 50:
        return "LOW"

    elif vehicles < 100:
        return "MODERATE"

    elif vehicles < 150:
        return "HIGH"

    else:
        return "VERY HIGH"


# ============================================================
# Step 17: Signal Time Recommendation
# ============================================================

def get_signal_time(vehicles):

    if vehicles < 50:
        return 20

    elif vehicles < 100:
        return 35

    elif vehicles < 150:
        return 50

    else:
        return 60


# ============================================================
# Step 18: Test One Prediction
# ============================================================

sample_input = pd.DataFrame({

    "Hour": [10],

    "Junction": [1],

    "Year": [2026],

    "Month": [10],

    "Day": [7],

    "DayOfWeek": [2],

    "Weekend": [0]
})


prediction = model.predict(sample_input)

predicted_vehicles = round(prediction[0])


congestion = get_congestion_level(
    predicted_vehicles
)


signal_time = get_signal_time(
    predicted_vehicles
)


print("\n====================================")
print("SAMPLE TRAFFIC PREDICTION")
print("====================================")

print("Predicted Vehicles       :", predicted_vehicles)

print("Congestion Level         :", congestion)

print("Recommended Green Signal :", signal_time, "seconds")


# ============================================================
# Step 19: Feature Importance
# ============================================================

importance = pd.DataFrame({

    "Feature": features,

    "Importance": model.feature_importances_

})


importance = importance.sort_values(

    by="Importance",

    ascending=False

)


print("\n====================================")
print("FEATURE IMPORTANCE")
print("====================================")

print(importance)


# ============================================================
# Step 20: Save Model
# ============================================================

model_path = r"C:\Users\ELCOT\Downloads\Mona\traffic_model.pkl"


joblib.dump(

    model,

    model_path

)


print("\n====================================")
print("MODEL SAVED")
print("====================================")

print("Model saved successfully!")

print("Location:")

print(model_path)


# ============================================================
# FINAL MESSAGE
# ============================================================

print("\n====================================")
print("TRAFFIC ML PROJECT COMPLETED")
print("====================================")

print("Model          : Random Forest Regressor")

print("Target         : Vehicles")

print("Features       :", features)

print("Prediction     : Vehicle Count")

print("Congestion     : Low / Moderate / High / Very High")

print("Signal Time    : Recommended Green Signal Duration")
```
